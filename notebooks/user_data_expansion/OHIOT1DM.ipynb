# OHIOT1DM User Data Expansion

This notebook creates user data expansion for the OHIOT1DM dataset following the same pattern as other diabetes clinical trial datasets (DCLP3, DCLP5, FLAIR, IOBP2, etc.).

## Dataset Overview
- **Study**: Ohio T1DM Dataset
- **Participants**: Starting with 12 participants
- **IDs**: 540, 544, 552, 567, 584, 596, 559, 563, 570, 575, 588, 591

## Standard Columns
Following the established pattern from other datasets:
- `id`: Participant identifier
- `insulin_delivery_device`: Device used for insulin delivery
- `insulin_delivery_algorithm`: Algorithm used for automated insulin delivery
- `cgm_device`: Continuous glucose monitoring device
- `ethnicity`: Participant ethnicity
- `age_of_diagnosis`: Age at diabetes diagnosis
- `is_pregnant`: Pregnancy status
- `insulin_delivery_modality`: Type of insulin delivery (AID, SAP, etc.)
- `insulin_type_bolus`: Insulin type used for bolus doses
- `insulin_type_basal`: Insulin type used for basal doses

```python
import pandas as pd
import numpy as np
import os
```

```python
# Create the base DataFrame with participant IDs
participant_ids = [540, 544, 552, 567, 584, 596, 559, 563, 570, 575, 588, 591]

# Initialize DataFrame with standard columns
df_ohiot1dm = pd.DataFrame({
    'id': participant_ids,
    'insulin_delivery_device': np.nan,
    'insulin_delivery_algorithm': np.nan, 
    'cgm_device': np.nan,
    'ethnicity': np.nan,
    'age_of_diagnosis': np.nan,
    'is_pregnant': False,  # Default to False unless specified
    'insulin_delivery_modality': np.nan,
    'insulin_type_bolus': np.nan,
    'insulin_type_basal': np.nan
})

print(f"Created OHIOT1DM DataFrame with {len(df_ohiot1dm)} participants")
print(f"Columns: {list(df_ohiot1dm.columns)}")
print("\nDataFrame preview:")
df_ohiot1dm
```

## Data Validation

Validate the data entry and check for consistency with other datasets.

```python
# Data validation checks
print("=== OHIOT1DM Data Validation ===")
print(f"Total participants: {len(df_ohiot1dm)}")
print(f"Unique IDs: {df_ohiot1dm['id'].nunique()} (should equal total participants)")
print(f"ID range: {df_ohiot1dm['id'].min()} - {df_ohiot1dm['id'].max()}")

print("\n=== Column Completeness ===")
for col in df_ohiot1dm.columns:
    non_null_count = df_ohiot1dm[col].notna().sum()
    percentage = (non_null_count / len(df_ohiot1dm)) * 100
    print(f"{col:25s}: {non_null_count:2d}/{len(df_ohiot1dm)} ({percentage:5.1f}%)")

print("\n=== Value Distributions ===")
categorical_cols = ['insulin_delivery_device', 'insulin_delivery_algorithm', 'cgm_device', 
                   'ethnicity', 'insulin_delivery_modality', 'insulin_type_bolus', 'insulin_type_basal']

for col in categorical_cols:
    if df_ohiot1dm[col].notna().any():
        print(f"\n{col}:")
        value_counts = df_ohiot1dm[col].value_counts(dropna=False)
        for value, count in value_counts.items():
            print(f"  {str(value):20s}: {count}")
    else:
        print(f"\n{col}: All values are NaN")
```

## Export Data

Save the OHIOT1DM user data expansion to CSV file.

```python
# Create output directory if it doesn't exist
output_dir = '../../data/user_data_expansion/'
os.makedirs(output_dir, exist_ok=True)

# Save to CSV
output_file = os.path.join(output_dir, 'OhioT1DM.csv')
df_ohiot1dm.to_csv(output_file, index=False)

print(f"✅ OHIOT1DM user data expansion saved to: {output_file}")
print(f"📊 Exported {len(df_ohiot1dm)} participant records")
print(f"📝 Columns: {len(df_ohiot1dm.columns)}")

# Verify file was saved
if os.path.exists(output_file):
    file_size = os.path.getsize(output_file)
    print(f"📁 File size: {file_size} bytes")
else:
    print("❌ Error: File was not saved successfully")
```
