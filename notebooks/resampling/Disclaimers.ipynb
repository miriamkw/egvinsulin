## Disclaimers and To Dos for next iteration

- Insulin types:
    - Some patients on MDI might have been included in the resampled data, because we took the first insulin type sample and used that for computing ice / iob for all insulin injections. In the future we need to either filter out MDI patients completely, or be clear about which insulin types are used for boluses and basals at each time step
    - Some patients might change insulin type throughout the study, so we should handle that approperiately
- Age: We use consistently the age at enrollment, but we should adjust the age throughout the dataset for subjects with more data than one year
- Weight and height: If there are several measurements, adjust thrughout the study
- Ice and iob: when we compute the total daily doses, we must ensure that the 0.0 and nan values are approperiately added. In this iteration, total daily dose might sometimes be too low if there are long periods in-between with missing data, where basal and bolus is set to 0.0. There should be a max limit for consequtive 0.0 values for basals, and boluses should always be set to nan instead of 0.0.
